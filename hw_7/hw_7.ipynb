### Парсинг материалов с сайта "Лайфхакер"

![separator1](https://i.imgur.com/ZUWYTii.png)

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from tqdm import tqdm
```

```python
base_url = 'https://lifehacker.ru/topics/technology/' # базовая часть ссылки, позже в цикле будем добавлять к ней пагинацию
response = requests.get('https://lifehacker.ru/topics/technology/') # получаем контент первой страиниц
soup = BeautifulSoup(response.text, 'lxml') # инициализируем объект bs4 и задаем парсер lxml
```

#### Получаем ссылки на статьи с первой страницы раздела "Технологии"

```python
raw_items = soup.find_all('a', class_="lh-small-article-card__link") # ищем все a-элементы с классом "lh-small-article-card__link", т.е. ссылки на статьи
links = [item.get('href') for item in raw_items] # получаем у ссылок только href-атрибут
links
```

#### Получаем ссылки на статьи с первых десяти страниц раздела "Технологии"

```python
parsed_urls = []

for page_num in range(1, 11):
    url = f'{base_url}?page={page_num}' # подставляем нужный номер страницы для пагинации, дальше парсим каждую страницу аналогично коду выше
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'lxml')
    raw_items = soup.find_all('a', class_='lh-small-article-card__link')
    parsed_urls += [item.get('href') for item in raw_items]
```

```python
parsed_urls
```

#### Парсим заголовки и тексты с первых десяти страниц раздела "Технологии"

```python
result = []

for url in tqdm(parsed_urls):
    article = {}

    if url.startswith('http'):
        article_url = url
    else:
        article_url = f'https://lifehacker.ru{url}'
        
    response = requests.get(article_url)
    soup = BeautifulSoup(response.text, 'lxml')

    try:
        article['title'] = soup.find('h1', class_='article-card__title').text.strip()
        article['text'] = soup.find('article', class_='single-article__post-content single-article__content-container').text.strip()
        result.append(article)
    except AttributeError:
        print(f"Не удалось распарсить {article_url}")
```
